```python

from nnunetv2.training.nnUNetTrainer.customized.unet_da import PlainConvUNet_DA
```

```python
from batchgenerators.utilities.file_and_folder_operations import load_json

plans_path = '/content/DA_models/Dataset139_BraTS2023/nnUNetTrainerNoDeepSupervision__nnUNetPlans__3d_fullres/plans.json'
plans = load_json(plans_path)
```

```python
config = plans['configurations']['3d_fullres']

input_channels = 4  # چون معمولا MRI با 4 modality هست.
n_stages = len(config['pool_op_kernel_sizes'])  # تعداد stages = تعداد resolution levels
features_per_stage = [min(config['UNet_base_num_features'] * (2 ** i), config['unet_max_num_features']) for i in range(n_stages)]
conv_op = torch.nn.Conv3d
kernel_sizes = config['conv_kernel_sizes']
strides = config['pool_op_kernel_sizes']
n_conv_per_stage = config['n_conv_per_stage_encoder']
n_conv_per_stage_decoder = config['n_conv_per_stage_decoder']
num_classes = 4  # در BraTS معمولا ۴ کلاس segmentation داریم (background, edema, necrosis, enhancing)
```

```python
state_dict_path = '/content/DA_models/Dataset139_BraTS2023/nnUNetTrainerNoDeepSupervision__nnUNetPlans__3d_fullres/fold_all/checkpoint_final.pth'
state_dict = torch.load(state_dict_path, map_location='cpu', weights_only=False)
```

```python
num_classes = 3  # برای match با مدل baseline

model = PlainConvUNet_DA(
    input_channels=input_channels,
    n_stages=n_stages,
    alpha=0.5,
    features_per_stage=features_per_stage,
    conv_op=conv_op,
    kernel_sizes=kernel_sizes,
    strides=strides,
    n_conv_per_stage=n_conv_per_stage,
    num_classes=num_classes,  # همینجا تغییر بده
    n_conv_per_stage_decoder=n_conv_per_stage_decoder,
    conv_bias=True,
    norm_op=torch.nn.InstanceNorm3d,
    norm_op_kwargs=None,
    dropout_op=None,
    dropout_op_kwargs=None,
    nonlin=torch.nn.LeakyReLU,
    nonlin_kwargs={'negative_slope': 0.01},
    deep_supervision=False,
    nonlin_first=False
)

# سپس:
model.load_state_dict(state_dict, strict=False)
model.eval()
print("✅ مدل با موفقیت لود شد (با استفاده از strict=False).")
```

```python
import torch.nn.utils.prune as prune

def apply_structured_pruning(model, amount=0.3):
    for name, module in model.named_modules():
        if isinstance(module, torch.nn.Conv3d):
            prune.ln_structured(module, name='weight', amount=amount, n=2, dim=0)
    return model

pruned_model = apply_structured_pruning(model, amount=0.3)
print("✅ Structured Pruning اعمال شد.")
```

```python
for name, module in pruned_model.named_modules():
    if isinstance(module, torch.nn.Conv3d):
        prune.remove(module, 'weight')

print("✅ ماسک‌های pruning حذف شدند (Permanent pruning).")
```

Structured pruning یعنی pruning فیلترهای کانولوشنی کامل (در dim=0) که باعث میشه مدل سبک‌تر بشه و معمولاً بهینه‌تر روی GPU اجرا بشه.

```python
total_filters = 0
pruned_filters = 0

for name, module in model.named_modules():
    if isinstance(module, torch.nn.Conv3d):
        total_filters += module.weight.shape[0]
        # mask وجود نداره چون permanent prune شده، اما می‌تونیم وزن‌های صفرشده رو بشماریم:
        filter_norms = torch.norm(module.weight.view(module.weight.shape[0], -1), dim=1)
        pruned = (filter_norms == 0).sum().item()
        pruned_filters += pruned
        print(f"Layer: {name} - Total filters: {module.weight.shape[0]}, Pruned filters: {pruned}")

print(f"\n🔍 Total filters: {total_filters}, Total pruned filters: {pruned_filters}")
print(f"🔍 Pruning ratio: {pruned_filters / total_filters * 100:.2f}%")
```

```python
'''  .
نوع: Structured Pruning
این یعنی به‌جای اینکه صرفاً مقادیر کوچک وزن‌ها رو صفر کنیم (unstructured)، ما کل فیلترها (یعنی بعد خروجی کانولوشن) رو حذف کردیم.

ابزار استفاده شده:

torch.nn.utils.prune.ln_structured
این تابع از PyTorch:

structured pruning انجام می‌ده.

L2-norm فیلترها رو محاسبه می‌کنه (magnitude-based pruning).

بر اساس amount=0.3، حدود ۳۰٪ از فیلترها رو (کم‌اثرترین‌ها) در dim=0 صفر کرده.

یعنی هر کانال خروجی (همون فیلتر) در tensor:

[out_channels, in_channels, D, H, W]
کامل حذف (zero) شده.


الگوریتم pruning:
L2-Norm structured pruning:

برای هر فیلتر (out_channel)، وزن‌های مربوطه رو flatten کرده و L2-norm گرفته.

پایین‌ترین ۳۰٪ فیلترها رو انتخاب کرده و mask کرده.

سپس با prune.remove این فیلترها permanent حذف شدند.

 کجا اعمال شد؟
روی همه‌ی لایه‌های Conv3d که در مدل DA-Unet تعریف شده بودن.

در encoder، decoder و حتی domain_classifier.

segmentation head هم (به‌صورت ناخواسته) prune شد.

مزایا و معایب:
✅ مزیت:

مدل سبک‌تر می‌شه و در inference سریع‌تر اجرا می‌شه.

structured pruning روی GPU نسبت به unstructured pruning بهتر بهینه‌سازی می‌شه.

❌ عیب:

pruning segmentation head باعث افت performance می‌شه.

pruning domain classifier ممکنه اثر منفی روی انتقال یادگیری بذاره.


'''
```

```python


# مسیر plans.json
plans_path = '/content/DA_models/Dataset139_BraTS2023/nnUNetTrainerNoDeepSupervision__nnUNetPlans__3d_fullres/plans.json'

# بارگذاری plans
with open(plans_path, 'r') as f:
    plans = json.load(f)

# تعداد کانال‌های ورودی و خروجی را مشخص می‌کنیم
in_channels = len(plans['modalities']) if 'modalities' in plans else 4
out_channels = len(plans['all_classes']) if 'all_classes' in plans else 4
```

مدل را دوباره می سازیم

```python
from dynamic_network_architectures.architectures.unet import PlainConvUNet
import torch.nn as nn

cfg = plans['configurations']['3d_fullres']
n_stages = len(cfg['pool_op_kernel_sizes']) + 1

# Extend encoder stages
n_conv_per_stage_encoder = cfg['n_conv_per_stage_encoder']
if len(n_conv_per_stage_encoder) < n_stages:
    n_conv_per_stage_encoder = n_conv_per_stage_encoder + [n_conv_per_stage_encoder[-1]]

# Extend decoder stages
n_conv_per_stage_decoder = cfg['n_conv_per_stage_decoder']
if len(n_conv_per_stage_decoder) < (n_stages - 1):
    n_conv_per_stage_decoder = n_conv_per_stage_decoder + [n_conv_per_stage_decoder[-1]]

# Extend kernel sizes
kernel_sizes = cfg['conv_kernel_sizes']
if len(kernel_sizes) < n_stages:
    kernel_sizes = kernel_sizes + [kernel_sizes[-1]]

# Extend strides
strides = cfg['pool_op_kernel_sizes']
if len(strides) < n_stages:
    strides = strides + [strides[-1]]

# Build feature maps per stage
features_per_stage = []
base = cfg['UNet_base_num_features']
for s in range(n_stages):
    next_val = min(base * (2 ** s), cfg['unet_max_num_features'])
    features_per_stage.append(next_val)

# Build model
model = PlainConvUNet(
    input_channels=in_channels,
    num_classes=out_channels,
    n_stages=n_stages,
    features_per_stage=features_per_stage,
    conv_op=nn.Conv3d,
    kernel_sizes=kernel_sizes,
    strides=strides,
    n_conv_per_stage=n_conv_per_stage_encoder,
    n_conv_per_stage_decoder=n_conv_per_stage_decoder,
    deep_supervision=False
)

# Print model structure
print(model)
```
