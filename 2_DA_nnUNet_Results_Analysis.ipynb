```python
import os
import shutil
import glob

source_dir = "/content/BraTS_PEDs_2023/ASNR-MICCAI-BraTS2023-PED-Challenge-TrainingData"
labels_dir = "/content/labelsTs"
os.makedirs(labels_dir, exist_ok=True)

cases = sorted(os.listdir(source_dir))
for case in cases:
    case_path = os.path.join(source_dir, case)
    if not os.path.isdir(case_path):
        continue
    seg_file = glob.glob(os.path.join(case_path, "*seg.nii.gz"))
    if len(seg_file) == 1:
        shutil.copy(seg_file[0], os.path.join(labels_dir, f"{case}.nii.gz"))
    else:
        print(f"Segmentation missing for case: {case}")
```

بریم سراغ مشاهده نتایج

```python
import os
import json
import pandas as pd

# مسیر پوشه‌ها
base_dir = '/content/drive/MyDrive/DA_nnUNet_Results'
folders = [
    'AllFolds_NoTTA',
    'AllFolds_TTA',
    'pred_fold0_TTA',
    'pred_fold0_noTTA'
]

results = []

for folder in folders:
    folder_path = os.path.join(base_dir, folder)
    if 'AllFolds' in folder:
        for fold_num in range(5):
            summary_path = os.path.join(folder_path, f'fold_{fold_num}', 'summary.json')
            if os.path.exists(summary_path):
                with open(summary_path, 'r') as f:
                    summary_data = json.load(f)
                dice_fg = summary_data.get('foreground_mean', {}).get('Dice', 'N/A')
                dice_123 = summary_data.get('mean', {}).get('(1, 2, 3)', {}).get('Dice', 'N/A')
                dice_23 = summary_data.get('mean', {}).get('(2, 3)', {}).get('Dice', 'N/A')
                dice_3 = summary_data.get('mean', {}).get('(3,)', {}).get('Dice', 'N/A')
                hd_fg = summary_data.get('foreground_mean', {}).get('Hausdorff95', 'N/A')
                hd_123 = summary_data.get('mean', {}).get('(1, 2, 3)', {}).get('Hausdorff95', 'N/A')
                results.append({
                    'Folder': folder,
                    'Fold': fold_num,
                    'Dice_foreground': dice_fg,
                    'Dice_1_2_3': dice_123,
                    'Dice_2_3': dice_23,
                    'Dice_3': dice_3,
                    'HD95_foreground': hd_fg,
                    'HD95_1_2_3': hd_123
                })
    else:
        summary_path = os.path.join(folder_path, 'summary.json')
        if os.path.exists(summary_path):
            with open(summary_path, 'r') as f:
                summary_data = json.load(f)
            dice_fg = summary_data.get('foreground_mean', {}).get('Dice', 'N/A')
            dice_123 = summary_data.get('mean', {}).get('(1, 2, 3)', {}).get('Dice', 'N/A')
            dice_23 = summary_data.get('mean', {}).get('(2, 3)', {}).get('Dice', 'N/A')
            dice_3 = summary_data.get('mean', {}).get('(3,)', {}).get('Dice', 'N/A')
            hd_fg = summary_data.get('foreground_mean', {}).get('Hausdorff95', 'N/A')
            hd_123 = summary_data.get('mean', {}).get('(1, 2, 3)', {}).get('Hausdorff95', 'N/A')
            results.append({
                'Folder': folder,
                'Fold': 'N/A',
                'Dice_foreground': dice_fg,
                'Dice_1_2_3': dice_123,
                'Dice_2_3': dice_23,
                'Dice_3': dice_3,
                'HD95_foreground': hd_fg,
                'HD95_1_2_3': hd_123
            })

# DataFrame
df_results = pd.DataFrame(results)
df_results.sort_values(['Folder', 'Fold'], inplace=True)

# تغییر نام ستون‌ها
df_results.rename(columns={
    'Dice_3': 'mean_DSC_ET',
    'Dice_2_3': 'mean_DSC_TC',
    'Dice_1_2_3': 'mean_DSC_WT',
    'HD95_3': 'mean_HD95_ET',
    'HD95_2_3': 'mean_HD95_TC',
    'HD95_1_2_3': 'mean_HD95_WT'
}, inplace=True)

# مرتب‌سازی ستون‌ها (فقط ستون‌های موجود)
columns_order = [
    'Folder', 'Fold',
    'mean_DSC_ET', 'mean_DSC_TC', 'mean_DSC_WT',
    'mean_HD95_ET', 'mean_HD95_TC', 'mean_HD95_WT'
]
existing_columns = [col for col in columns_order if col in df_results.columns]

df_final = df_results[existing_columns]

# نمایش نتایج نهایی
from IPython.display import display
display(df_final)
```

```python
# ستون‌های mean DSC
df_results['mean_DSC_ET'] = df_results['mean_DSC_ET']
df_results['mean_DSC_TC'] = df_results['mean_DSC_TC']
df_results['mean_DSC_WT'] = df_results['mean_DSC_WT']

# چون mean_HD95_ET و بقیه رو نداریم، با N/A پر کنیم
df_results['mean_HD95_ET'] = 'N/A'
df_results['mean_HD95_TC'] = 'N/A'
df_results['mean_HD95_WT'] = 'N/A'

# مرتب‌سازی به سبک جدول مقاله
columns_order = [
    'Folder', 'Fold',
    'mean_DSC_ET', 'mean_DSC_TC', 'mean_DSC_WT',
    'mean_HD95_ET', 'mean_HD95_TC', 'mean_HD95_WT'
]

df_final = df_results[columns_order]

# نمایش
from IPython.display import display
display(df_final)
```
